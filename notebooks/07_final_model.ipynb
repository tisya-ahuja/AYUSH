## Imports

```python
import sys
print(sys.executable)
import pandas as pd
import numpy as np
import joblib

from xgboost import XGBRegressor
```

## Load the data

```python
train = pd.read_csv(
    "../data/processed/FD001/train_FD001_with_rul.csv"
)

print("Shape:", train.shape)
display(train.head())
```

## recreating the baseline feature set

```python
feature_columns = [
    "cycle",
    "op_setting_1",
    "op_setting_2",
    "sensor_2",
    "sensor_3",
    "sensor_4",
    "sensor_6",
    "sensor_7",
    "sensor_8",
    "sensor_9",
    "sensor_11",
    "sensor_12",
    "sensor_13",
    "sensor_14",
    "sensor_15",
    "sensor_17",
    "sensor_20",
    "sensor_21"
]

X = train[feature_columns].copy()
y = train["RUL"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Number of features:", len(feature_columns))
```

## Training the Baseline XGBoost

```python
final_model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

final_model.fit(X, y)

print("Final XGBoost model trained.")
```

```python
model_path = "../models/fd001_final_model.pkl"

joblib.dump(
    final_model,
    model_path
)

print("Saved:", model_path)


pd.DataFrame({
    "feature": feature_columns
}).to_csv(
    "../reports/FD001/final_features.csv",
    index=False
)

print("Saved: ../reports/FD001/final_features.csv")
```
