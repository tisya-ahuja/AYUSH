## Column Naming and Cleaning

```python
columns = (
    ["unit", "cycle"]
    + [f"op_setting_{i}" for i in range(1, 4)]
    + [f"sensor_{i}" for i in range(1, 22)]
)

print("Number of columns:", len(columns))
print(columns)
```

```python
train = train_raw.copy()
train.columns = columns

display(train.head())
print(train.shape)
print(train.columns.tolist())
```

```python
print(train.dtypes)
```

```python
print("Total missing values:", train.isna().sum().sum())
print("Duplicate rows:", train.duplicated().sum())
constant_columns = train.columns[
    train.nunique() <= 1
]

print("Constant columns:")
print(constant_columns.tolist())
```

```python
train_clean = train.drop(columns=constant_columns).copy()

print("Original shape:", train.shape)
print("Cleaned shape:", train_clean.shape)

display(train_clean.head())
```

```python
train_clean.to_csv('../data/processed/FD001/train_FD001_clean.csv',index=False)

print("Saved: ../data/processed/FD001/train_FD001_clean.csv")
```
