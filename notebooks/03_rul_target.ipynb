## Imports

```python
import sys
print(sys.executable)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

## Load the data

```python
train_clean = pd.read_csv(
    "../data/processed/FD001/train_FD001_clean.csv"
)

display(train_clean.head())
print(train_clean.shape)
```

## Create RUL

```python
max_cycle = train_clean.groupby("unit")["cycle"].transform("max")

train_clean["RUL"] = max_cycle - train_clean["cycle"]

display(train_clean[["unit", "cycle", "RUL"]].head(10))
print("RUL min:", train_clean["RUL"].min())
print("RUL max:", train_clean["RUL"].max())
print("Missing RUL:", train_clean["RUL"].isna().sum())
```

```python
last_cycles = train_clean.loc[
    train_clean.groupby("unit")["cycle"].idxmax(),
    ["unit", "cycle", "RUL"]
]

display(last_cycles.head(10))

print(
    "Engines with final RUL not equal to 0:",
    (last_cycles["RUL"] != 0).sum()
)
```

```python
train_clean.to_csv(
    "../data/processed/FD001/train_FD001_with_rul.csv",
    index=False
)

print("Saved: ../data/processed/FD001/train_FD001_with_rul.csv")
```
